# Data cleaning

It seems that we are unable to read the picklefile. And there are no other descriptions/explanations in the folders that explain which features in the dataset (data.npy) is which feature. However, based on the provided images 0.png, 1.png etc., and plotting them ourselves as done above, we can assume that the structure of the features $X$ remains the same as before. And hence, the first six variables are $X$ with the same structure as before, in the older datasets and as described in the getting_started.ipynb in the gitlab repo of AdvAI. However, before the forcing features $Y$ were present in the data["forcing"] section of the npz file, they are no longer. So, it is unclear which of the last 6 variables are the 2 forcing variables. For now, we assume the 7th and 8th variable, however, we have to double check this.

```python
# ============================================
# CEL 1: Imports & paden instellen
# ============================================

import os
import shutil
from glob import glob

# Pad naar de ruwe data:
RAW_DATA_DIR = "/home/ubuntu/data/TCV150683_database_dynamic_density_ramps"

# Paden waar we de getransformeerde data willen opslaan
# (bijvoorbeeld train/val/test splits):
SAVE_ROOT = "/home/ubuntu/data/processed/"
TRAIN_DIR = os.path.join(SAVE_ROOT, "train")
VAL_DIR   = os.path.join(SAVE_ROOT, "val")
TEST_DIR  = os.path.join(SAVE_ROOT, "test")

# Maak deze directories (indien ze nog niet bestaan):
os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(VAL_DIR, exist_ok=True)
os.makedirs(TEST_DIR, exist_ok=True)
```

```python
# ============================================
# CEL 2: Functie om data te converteren
# ============================================

def convert_raw_npy_to_npz(raw_npy_path, npz_save_path):
    """
    Leest de ruwe data.npy in (shape: (T, 500, 12)) en selecteert:
      - X = eerste 6 features  (index: 0..5)
      - Y = 7e en 8e feature   (index: 6..7)

    Slaat deze daarna op in een .npz-file met keys "output" en "forcing",
    zodat de FusionDataset-class ze direct kan inladen.
    """
    data = np.load(raw_npy_path)   # data.shape = (T, 500, 12)

    # Check shape en aantal features
    if data.shape[-1] < 8:
        raise ValueError(f"Minimaal 8 features nodig, maar data heeft shape: {data.shape}")

    # X = de eerste 6 features
    X = data[..., :6]   # shape (T, 500, 6)
    # Y = de 7e en 8e features
    Y = data[..., 6:8]  # shape (T, 500, 2)

    # Schrijf weg als .npz
    # 'output' -> X, 'forcing' -> Y
    np.savez(npz_save_path, output=X, forcing=Y)
```

```python
# ============================================
# CEL 3: Train/Val/Test split en conversie
# ============================================

# Zoek alle subdirectories in de RAW_DATA_DIR waarin een data.npy staat.
# Elke subdir wordt gezien als 1 "traject" uit de simulatie.
subdirs = []
for root, dirs, files in os.walk(RAW_DATA_DIR):
    # Kijk of 'data.npy' in de bestanden zit:
    if "data.npy" in files:
        subdirs.append(root)

# Sorteer en shuffle (optioneel) zodat de split random gebeurt
subdirs = sorted(subdirs)
np.random.shuffle(subdirs)

# Simpele splits (bijv. 60% train, 20% val, 20% test)
n_total = len(subdirs)
n_train = int(0.6 * n_total)
n_val   = int(0.2 * n_total)
# de rest is test
n_test  = n_total - n_train - n_val

train_subdirs = subdirs[:n_train]
val_subdirs   = subdirs[n_train:n_train + n_val]
test_subdirs  = subdirs[n_train + n_val:]

print(f"Totaal  : {n_total} trajectories gevonden")
print(f"Train   : {len(train_subdirs)}")
print(f"Val     : {len(val_subdirs)}")
print(f"Test    : {len(test_subdirs)}")

def process_and_save(subdir_list, save_dir):
    """
    Leest per subdir de 'data.npy' in, converteert deze met
    convert_raw_npy_to_npz() en slaat op in save_dir.
    """
    for i, sd in enumerate(subdir_list):
        # pad naar de ruwe data.npy
        raw_npy = os.path.join(sd, "data.npy")

        # maak een logische naam voor de npz-file
        # bijvoorbeeld 'traj_0001.npz', 'traj_0002.npz', etc.
        filename = f"traj_{i:04d}.npz"
        save_path = os.path.join(save_dir, filename)

        # converteren en opslaan
        convert_raw_npy_to_npz(raw_npy, save_path)

# Voer nu de functie uit voor train/val/test
process_and_save(train_subdirs, TRAIN_DIR)
process_and_save(val_subdirs,   VAL_DIR)
process_and_save(test_subdirs,  TEST_DIR)

print("Conversie gereed!")
```
